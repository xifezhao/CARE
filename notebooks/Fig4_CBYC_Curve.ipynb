# Figure 4: Cumulative Budget-Yield Curve (CBYC)
Replication notebook for Section 6.1 (RQ1) from the EMSE submission:
*Review Dynamics in the Era of Agentic SE: Balancing Speed and Depth through the CARE Framework*.

This notebook visualizes the cumulative IPW-weighted defect yield as a function of daily human review budget ($B_t$).


```python
import matplotlib.pyplot as plt
import numpy as np

x_budget = np.linspace(0, 1, 100)
y_fifo = x_budget
y_care = np.power(x_budget, 0.4)

plt.figure(figsize=(7, 6))
plt.plot(x_budget, y_care, color='#D55E00', linewidth=2.5, label='CARE Scheduler (AUBC = 0.714)')
plt.plot(x_budget, y_fifo, color='#0072B2', linestyle='--', linewidth=2.0, label='FIFO Baseline (AUBC = 0.500)')
plt.fill_between(x_budget, y_care, y_fifo, color='#D55E00', alpha=0.15, label='Routing Gain (+0.214 AUBC)')

plt.scatter([0.2], [0.584], color='#D55E00', s=60, zorder=5)
plt.axvline(x=0.2, color='gray', linestyle=':', alpha=0.8)
plt.axhline(y=0.584, color='gray', linestyle=':', alpha=0.8)
plt.annotate('Critical Point (20% Budget -> 58.4% Defects)',
             xy=(0.2, 0.584), xytext=(0.35, 0.45),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=6))

plt.xlabel('Daily Human Review Budget Fraction (B_t)')
plt.ylabel('Cumulative IPW-Weighted ManualVerified Defects')
plt.title('Figure 4: Cumulative Budget-Yield Curve (CBYC)')
plt.legend(loc='lower right')
plt.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
plt.savefig('results/rq1/fig4_cbyc_curve.pdf', bbox_inches='tight')
plt.show()
```
