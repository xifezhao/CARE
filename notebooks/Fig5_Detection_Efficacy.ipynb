# Figure 5: Detection Efficacy Across Defect Categories
Replication notebook for Section 6.2 (RQ2).


```python
import matplotlib.pyplot as plt
import numpy as np

categories = ['Syntax/Style', 'Race Conditions', 'Auth Flows', 'Business Logic']
monoculture_rates = [92, 28, 34, 40]
orthogonal_rates = [95, 76, 82, 61]

x = np.arange(len(categories))
width = 0.35

plt.figure(figsize=(9, 5.5))
plt.bar(x - width/2, monoculture_rates, width, label='Monoculture (AI-on-AI)', color='#999999')
plt.bar(x + width/2, orthogonal_rates, width, label='Orthogonal Pipeline (CARE Phase 3)', color='#0072B2')

plt.ylabel('Detection Efficacy (%)')
plt.title('Figure 5: Vulnerability Detection Rates by Methodology')
plt.xticks(x, categories)
plt.ylim(0, 115)
for i in range(len(categories)):
    plt.text(x[i] - width/2, monoculture_rates[i] + 2, f'{monoculture_rates[i]}%', ha='center', fontsize=9)
    plt.text(x[i] + width/2, orthogonal_rates[i] + 2, f'{orthogonal_rates[i]}%', ha='center', fontsize=9)

plt.legend()
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()
```
