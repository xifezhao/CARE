# Figure 6: Automated Review Evaluation Trade-offs (RQ3)
Replication notebook for Section 6.3 (RQ3) from the EMSE submission:
*Review Dynamics in the Era of Agentic SE: Balancing Speed and Depth through the CARE Framework*.

This notebook visualizes the empirical performance across the five strictly controlled conditions:
- **Condition A**: Raw Context Baseline
- **Condition B**: Generic AI Summary
- **Condition C**: Information-Matched Unstructured Summary
- **Condition D**: CARE Evidence Pack (Advisory)
- **Condition D+Gate**: CARE Hard Compliance Gate

**Crucial Distinction**:
- Defect Localization Recall: Reviewer discovering and pinpointing the defect (90% for both D and D+Gate).
- Defect-PR Merge Interception: Preventing defect-bearing PRs from merging (90% for D vs. 100% for D+Gate).
- Unsupported Approvals: Erroneously approved defect PRs (10% for D vs. 0% for D+Gate).

```python
import matplotlib.pyplot as plt
import numpy as np

conditions = ['Cond A (Raw)', 'Cond B (Generic)', 'Cond C (Info-Match)', 'Cond D (Advisory)', 'Cond D+Gate (Hard)']
localization_recall = [0.133, 0.367, 0.700, 0.900, 0.900]
merge_interception = [0.133, 0.367, 0.700, 0.900, 1.000]
unsupported_approvals = [0.867, 0.633, 0.300, 0.100, 0.000]
misleading_resisted = [0.176, 0.235, 0.824, 0.941, 1.000]

x = np.arange(len(conditions))
width = 0.20

plt.figure(figsize=(12, 6))
plt.bar(x - 1.5*width, localization_recall, width, label='Defect Localization Recall (N=30)', color='#56B4E9')
plt.bar(x - 0.5*width, merge_interception, width, label='Defect-PR Merge Interception Rate (N=30)', color='#0072B2')
plt.bar(x + 0.5*width, unsupported_approvals, width, label='Unsupported Approval Rate (N=30)', color='#D55E00')
plt.bar(x + 1.5*width, misleading_resisted, width, label='Misleading Resistance Rate (N=17)', color='#009E73')

plt.ylabel('Rate (Fraction 0.0 - 1.0)', fontsize=12)
plt.title('Figure 6: Automated Review Performance Across Evidence Presentation and Policy Conditions', fontsize=13)
plt.xticks(x, conditions, fontsize=10)
plt.ylim(0, 1.20)
plt.legend(loc='upper right', framealpha=0.9, fontsize=9)
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.tight_layout()
plt.savefig('results/rq3/fig6_rq3_tradeoffs.pdf', bbox_inches='tight')
plt.show()
```
