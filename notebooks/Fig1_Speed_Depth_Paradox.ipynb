# Figure 1: The Speed-Depth Paradox in Agentic SE
Replication notebook for Figure 1 in EMSE submission.


```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
turnaround_traditional = np.random.exponential(scale=24.0, size=500) + 2.0
depth_traditional = np.clip(100 - 1.5 * turnaround_traditional + np.random.normal(0, 10, 500), 20, 95)

turnaround_agentic = np.random.exponential(scale=3.5, size=500) + 0.5
depth_agentic = np.clip(30 + 2.0 * turnaround_agentic + np.random.normal(0, 12, 500), 10, 75)

plt.figure(figsize=(8, 5))
plt.scatter(turnaround_traditional, depth_traditional, color='#0072B2', alpha=0.5, label='Human-Authored PRs')
plt.scatter(turnaround_agentic, depth_agentic, color='#D55E00', alpha=0.5, label='Agentic-PRs (AI-Assisted)')
plt.axvline(x=4.0, color='red', linestyle='--', alpha=0.7, label='Hyper-Fast Approval Threshold (<4h)')
plt.xlabel('Review Turnaround Time (Hours)')
plt.ylabel('Review Depth Score (Normalized)')
plt.title('Figure 1: The Speed-Depth Paradox in Code Review')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()
```
